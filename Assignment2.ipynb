```python
import pandas as pd
import numpy as np
import matplotlib.pylab as plt
import seaborn as sns 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv('Housing.csv')
```

```python
df.head()
```

```python
print(df.isnull().sum())
```

```python
# 2. Separate input and output
X = df[['area']]
y = df['price']
```

```python
# 3. Train the model
model = LinearRegression()
model.fit(X,y)
```

```python
# 4. Make predictions
y_pred = model.predict(X)
```

```python
# 5. Calculate slope and intercept
slope = model.coef_[0]
intercept = model.intercept_
```

```python
# 6. Calculate RMSE
RMSE = np.sqrt(mean_squared_error(y, y_pred))
```

```python
# 7. Display results
print("Slope (Coefficient):", slope)
print("Intercept:", intercept)
print("RMSE:", RMSE)
```

```python
plt.figure(figsize=(6,8))
plt.scatter(df['area'], df['price'])

plt.xlabel('price')
plt.ylabel('area')
plt.title('Scatter plot')
plt.grid()
plt.show()
```

```python
# Sort area values to draw a proper regression line
sorted_indices = X['area'].argsort()

plt.plot(
    X['area'].iloc[sorted_indices],
    y_pred[sorted_indices],
    
    label ='regression line',
    color ='red'
)

plt.title('Linear Regression: House Price vs Area')
plt.xlabel('Area')
plt.ylabel('Price')
plt.legend()
plt.grid()
plt.show()
```
