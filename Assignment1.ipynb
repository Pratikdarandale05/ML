```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns
```

```python
df = pd.read_csv("Titanic.csv")
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df.isnull().sum())
```

```python
df.drop('Cabin', axis=1, inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0], inplace = True)
print(df.isnull().sum())

```

```python
import matplotlib.pylab as plt
import seaborn as sns

```

```python
plt.figure(figsize=(8,6))
plt.boxplot(df['Age'])
plt.xlabel('Age')

plt.title('Age Distribution')
plt.show()
```

```python
plt.figure(figsize=(6,7))

plt.plot(df['Sex'], df['Survived'])

plt.xlabel('Age')
plt.ylabel('Sex')
plt.show()
```

```python
plt.figure(figsize=(6,8))
plt.bar(df['Sex'],df['Age'])
plt.title('Gender with age ')
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```

```python
plt.scatter(df['Age'], df['Pclass'])
plt.grid()
plt.show()
```

```python

gender = df['Sex'].value_counts()

plt.pie(gender, labels = gender.index, autopct = '%1.1f%%', startangle=90)
plt.title("Gender Distribution")

plt.show()
```
